```python
df['DateTime']= pd.to_datetime(df['DateTime'], utc=True)
df['Date']= pd.to_datetime(df['Date'])
```

```python
#Histogram

import matplotlib.pyplot as plt

plt.plot(df['DateTime'], df['Close'])
plt.title('Gold price over Time(2000-2026)')
plt.xlabel('Date')
plt.ylabel('Closing price (USD)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Gold Price Over Time (2000–2026) — Observations

### Overall Trend
Gold has shown a **strong long-term upward trend**, rising from **\$300 in 2000** to over **\$5000 in 2026** — a **~16x increase** over 26 years.

---

### Key Phases

| Period | Price Range | Event |
|--------|------------|-------|
| 2000–2008 | \$300 → \$900 | Slow steady rise, post dot-com era |
| 2008–2011 | \$900 → \$1800 | Sharp rally driven by the 2008 Global Financial Crisis — investors moved to gold as a safe haven |
| 2011–2015 | \$1800 → \$1100 | Peak followed by correction and consolidation |
| 2015–2019 | \$1100 → \$1300 | Relatively flat, low volatility period |
| 2020 | \$1300 → \$2000 | COVID-19 pandemic drove global uncertainty, gold spiked |
| 2022–2026 | \$2000 → \$5000+ | Explosive growth due to geopolitical tensions, inflation fears, and global instability |

---

### Key Takeaways
- Gold tends to **rise during crises** (financial, health, geopolitical)
- Long consolidation periods are followed by **sharp breakouts**
- The **2024–2026 surge** is the steepest in the dataset's history

```python
plt.figure(figsize=(12, 5))
plt.hist(df['Price_Change_Percent'], bins=100, color='steelblue', edgecolor='black', alpha=0.7)
plt.title('Distribution of Daily Price Change % (2000-2026)')
plt.xlabel('Price Change %')
plt.ylabel('Frequency')
plt.axvline(df['Price_Change_Percent'].mean(), color='red', linestyle='--', label=f"Mean: {df['Price_Change_Percent'].mean():.3f}%")
plt.axvline(0, color='black', linestyle='-', linewidth=1)
plt.legend()
plt.tight_layout()
plt.show()
```

## Distribution of Daily Price Change % (2000–2026) — Observations

### Shape
The distribution is **approximately bell-shaped (normal)**, 
centered very close to **0%**, which is typical for financial assets.

---

### Key Statistics
- **Mean: 0.052%** — slightly positive, confirming the long-term 
  upward trend
- Most daily moves fall between **-2.5% and +2.5%**
- Extreme moves beyond **±5%** are very rare

---

### Key Takeaways
- Gold is a **relatively stable asset** — most days see very small changes
- The slight **positive skew** (mean > 0) confirms gold's long-term 
  bullish trend
- The rare extreme values (beyond ±7.5%) represent 
  **major market shock days** like 2008 crisis or COVID-19

```python
monthly_avg = df.groupby('Month')['Close'].mean()

plt.figure(figsize=(10, 5))
plt.bar(monthly_avg.index, monthly_avg.values, color='gold', edgecolor='black', alpha=0.8)
plt.title('Average Gold Price by Month (2000-2026)')
plt.xlabel('Month')
plt.ylabel('Avg Close Price (USD)')
plt.xticks(range(1,13), ['Jan','Feb','Mar','Apr','May','Jun',
                          'Jul','Aug','Sep','Oct','Nov','Dec'])
plt.ylim(1150, 1350)
plt.tight_layout()
plt.show()
```

## Average Gold Price by Month (2000–2026) — Observations

### Seasonal Pattern
Gold prices show a **mild but consistent seasonal pattern** 
across the year.

---

### Key Observations
- **January** is the strongest month — highest average price (~\$1325)
- **March** is the weakest month — lowest average price (~\$1223)
- Prices **dip in spring** (Mar–Jun) and **recover in second half** 
  of the year
- **December** sees a year-end rise (~\$1298) — possibly due to

### Key Takeaways

```python
import seaborn as sns

corr_cols = ['Open','High','Low','Close','Volume',
             'Price_Change','Price_Change_Percent',
             'MA_7','MA_30','MA_90','MA_365','Volatility_30d']

plt.figure(figsize=(12, 8))
sns.heatmap(df[corr_cols].corr(), annot=True, fmt='.2f', 
            cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap (2000-2026)')
plt.tight_layout()
plt.show()
```

## Correlation Heatmap (2000–2026) — Observations

### Strong Positive Correlations 🔴
- **Open, High, Low, Close** are almost perfectly correlated 
  (~1.00) — expected, as they all move together daily
- **MA_7, MA_30, MA_90, MA_365** are highly correlated with 
  Close (~0.97–1.00) — moving averages follow the price closely
- **Price_Change and Price_Change_Percent** are correlated 
  (~0.6) — both measure daily movement

---

### Weak/No Correlations 🔵
- **Volume** has very low correlation with price (~0.04–0.06) 
  — high volume doesn't always mean high price
- **Volatility_30d** has low correlation with price (~0.06) 
  — volatility is independent of price level
- **Price_Change_Percent** has near-zero correlation with MAs 
  — daily % moves don't depend on long-term averages

---

### Key Takeaways
- OHLC prices are **redundant** with each other — for modelling, 
  just use `Close`
- Moving averages are **not independent features** — they're 
  derived from Close, so using all of them together in a model 
  would cause **multicollinearity**
- **Volume and Volatility** are the most **independent** features 
  — useful for modelling alongside Close

```python
plt.figure(figsize=(14, 6))
plt.plot(df['Date'], df['Close'], color='steelblue', alpha=0.5, linewidth=1, label='Close Price')
plt.plot(df['Date'], df['MA_7'], color='orange', linewidth=1.2, label='MA 7')
plt.plot(df['Date'], df['MA_30'], color='green', linewidth=1.2, label='MA 30')
plt.plot(df['Date'], df['MA_365'], color='red', linewidth=1.5, label='MA 365')
plt.title('Gold Price with Moving Averages (2000-2026)')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend()
plt.tight_layout()
plt.show()
```

## Gold Price with Moving Averages (2000–2026) — Observations

### What are Moving Averages?
Moving averages **smooth out daily price noise** to reveal the 
underlying trend. Shorter MAs react faster, longer MAs are smoother.

---

### Key Observations
- **MA_7 and MA_30** stay very close to the Close price — 
  they react quickly to price changes
- **MA_365** is much smoother — it represents the 
  long-term trend and lags behind sharp moves
- When Close price is **above MA_365** — gold is in an 
  **uptrend** (bullish signal)
- When Close price is **below MA_365** — gold is in a 
  **downtrend** (bearish signal) — visible during 2013–2018

---

### Key Takeaways
- MA_365 clearly shows **three major phases**:
  - Rising (2000–2012)
  - Declining/flat (2012–2019)
  - Explosive rise (2020–2026)
- The **gap between Close and MA_365 in 2025–2026** is the 
  largest ever — indicating an unusually strong bull run
- Moving averages are widely used in **trading strategies** 
  as buy/sell signals

```python
plt.figure(figsize=(14, 5))
plt.plot(df['Date'], df['Volatility_30d'], color='crimson', linewidth=1, alpha=0.8)
plt.fill_between(df['Date'], df['Volatility_30d'], alpha=0.3, color='crimson')
plt.title('30-Day Rolling Volatility (2000-2026)')
plt.xlabel('Date')
plt.ylabel('Volatility (%)')
plt.tight_layout()
plt.show()
```

## 30-Day Rolling Volatility (2000–2026) — Observations

### What is Volatility?
Volatility measures how much the price **fluctuates day to day**. 
High volatility = uncertain, fast-moving market. 
Low volatility = stable, calm market.

---

### Key Observations

| Period | Volatility | Cause |
|--------|-----------|-------|
| 2000–2007 | Low (~0.5–1.5%) | Stable, gradual growth period |
| 2008–2009 | Spike to ~3.2% | **Global Financial Crisis** |
| 2011–2013 | Elevated ~2–2.5% | Post-crisis uncertainty |
| 2014–2019 | Low ~0.5–1% | Calm consolidation phase |
| 2020 | Spike to ~2.6% | **COVID-19 pandemic** |
| 2025–2026 | Spike to ~3% | Geopolitical tensions, trade wars |

---

### Key Takeaways
- Volatility **spikes during crises** and returns to normal after
- The **2008 spike is the highest** in the dataset (~3.2%)
- Long calm periods (2014–2019) suggest gold was 
  **range-bound and predictable**
- Recent 2025–2026 spike confirms the current 
  **bull run is driven by uncertainty**
